2. Explain the difference between a for loop and a while loop in Python.

A for loop is used when the number of iterations are known. In other words, the range of iteration is known.    
Example: for i in [1, 2, 3, 4]  In this statement, the range is limited and known, as we know the length of the list. Hence it will take a maximum 4 iterations for the loop to come to conclusion.


Whereas a while loop is used when the range of iterations is unknown.   
Example: while i != 0   In this statement, the range is unknown and it can take any number of iterations for the pointer to come out of while loop.

## Recursion

1. Explain the concept of recursion in Python. How does it differ from iteration?

Recursion is a technique where a function calls itself in order to solve a problem. It is mainly used to solve problems that can be broken down into smaller versions of the same problem.

The technique of iteration mainly involves a program going over the same set of statements in a loop over a range or iterable. Thi differs from recursion wherein the loop involves the function calling itself again and again until the problem cannot be broken down further.

2. Write a Python program to calculate the factorial of a number using recursion.

```python
def fact(num):
  if num == 0 or num == 1:
    return 1
  factorial = num*fact(num-1)
  return factorial

print(fact(5))
```

3. Create a recursive Python function to find the nth Fibonacci number.


```python
def fib(num):
  if num == 1:
    return 0
  if num == 2:
    return 1
  fibonacci = fib(num-1)+fib(num-2)
  return fibonacci

print(fib(11))
```

6. Create a recursive Python function to find the greatest common divisor (GCD) of two numbers using the Euclidean algorithm.


```python
def gcd(a,b):
  if a > b:
    if b == 0:
      return a
    else:
      return gcd(b, a%b)
  else:
    if a == 0:
      return b
    else:
      return gcd(a, b%a)

gcd(72, 27)
```

8. Create a recursive Python function to calculate the power of a number (x^n).


```python
def power(x, n):
  if n == 0:
    return 1
  if n == 1:
    return x
  return x*power(x, n-1)

power(-3,4)
```

9. Write a recursive Python function to find all permutations of a given string.


```python
def string_permutations(s):
    if len(s) <= 1:
        return [s]
    permutations = []
    for i, char in enumerate(s):
        remaining = s[:i] + s[i+1:] # a
        for perm in string_permutations(remaining):
            permutations.append(char + perm)
    return permutations

string_permutations('abc')
```

11. Create a recursive Python function to generate all possible combinations of a list of elements.


```python
def generate_combinations(lst):
    if not lst:
        return [[]]

    first_element = lst[0]
    remaining_combinations = generate_combinations(lst[1:])

    combinations_with_first = [ [first_element] + combo for combo in remaining_combinations ]

    return remaining_combinations + combinations_with_first

generate_combinations([1,2,3])
```

2. Explain the concept of variable scope in Python, including local and global variables.

    Scope of a variable is the region in a program where the variable is accessible and can be used.

      1. Local Variables are declared within a function. Its accessibilty is limited to within that function only. Once the function exits, these variables are destroyed.

      2. Global Variables are declared outside any function or block, and these are accessible and used throughout the program. Modifying a global variable inside a function requires the use of the keyword `global`.

3. Write a Python program that demonstrates the use of global variables within functions.


```python
s = 20

def addition(a, b):
  global s
  print(a, b, s)
  print(a+b+s)
  s = 10
  return a+b+s

addition(10, 10)
```

4. Create a Python function that calculates the factorial of a number and returns it.


```python
def fact(num):
  if num == 0 or num == 1:
    return 1
  return num*fact(num-1)

fact(5)
```

3. Explain the concept of higher-order functions in Python, and provide an example.

    A higher-order function is a function that:

    1. Takes other functions as arguments.
    2. Returns a function as its result.

    Example:


```python
def apply(func, *value):
  return func(*value)

def square(s):
  return s*s

apply(square, 6)
```
