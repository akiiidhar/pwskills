Problem 1: Bank Account

```python
import logging

logging.basicConfig(filename=r'logs\BankAccount.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class BankAccount:

    logging.info('Entering class BankAccount.')

    def __init__(self, acc_num, acc_holder_name, acc_balance):
        try:
            logging.info('Entering __init__ method. Initialising variables: account_no, name & balance.')
            self.account_no = acc_num
            self.name = acc_holder_name
            self.balance = acc_balance
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def deposit(self, depo_amt):
        try:
            logging.info(f'Depositing to A/C No. {self.account_no}.')
            self.balance += depo_amt
            logging.info(f'An amount of {depo_amt} has been deposited to A/C No. {self.account_no}.\nNew balance is {self.balance}.')
            return self.balance
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
            
    def withdraw(self, wdrw_amt):
        try:
            logging.info(f'Withdrawing from A/C No. {self.account_no}.')
            self.balance -= wdrw_amt
            logging.info(f'An amount of {wdrw_amt} has been withdrawn from A/C No. {self.account_no}.\nNew balance is {self.balance}.')
            return self.balance
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

```python
import logging

logging.basicConfig(filename=r'logs\EmployeeManagement.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class Employee:
    logging.info('Entering class Employee.')    

    def __init__(self, eid, ename, esalary):
        try:
            logging.info('Entering __init__ method. Initialising variables: emp_ID, emp_name & emp_salary.')
            self.emp_ID = eid
            self.emp_name = ename
            self.emp_salary = esalary
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def yearly_bonus(self, bns_percent):
        try:
            logging.info('Calculating yearly bonus.')
            bonus = (self.emp_salary*bns_percent)/100
            logging.info(f'Yearly Bonus: {bonus}')
            return bonus
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def emp_details(self):
        try:
            logging.info(f'Displaying employee details...\nEmployee Name: {self.emp_name}\nEmployee ID: {self.emp_ID}\nEmployee Salary: {self.emp_salary}')
            print(f'Employee Name: {self.emp_name}\nEmployee ID: {self.emp_ID}\nEmployee Salary: {self.emp_salary}')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

```python
import logging

logging.basicConfig(filename=r'logs\VehicleRental.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class VehicleRental:

    logging.info('Entering class VehicleRental')

    def __init__(self):
        logging.info('Entering __init__ method. Declaring variable: avl_veh.')
        self.avl_veh = ['Chevrolet Astro', 'Volkswagen GLI', 'Mercedes Benz GLE Class', 'Honda Pilot', 'Chevrolet Avalanche']

    def display_vehicles(self):
        try:
            logging.info(f'Displaying available vehicles: {self.avl_veh}')
            return self.avl_veh
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def rent_vehicle(self, chosen_veh):
        try:
            logging.info(f'Checking if {chosen_veh} is available...')
            if chosen_veh in self.avl_veh:
                logging.info('Vehicle available for rent!!')
                choice = str(input(f'Do you want to rent {chosen_veh}? Enter "Yes" or "No": '))
                if choice.lower() == 'yes':
                    logging.info(f'{chosen_veh} rented!!')
                    self.avl_veh.remove(chosen_veh)
                    return f'{chosen_veh} rented!!'
                else:
                    logging.info(f'{chosen_veh} not rented!!')
                    return f'{chosen_veh} not rented!!'
            else:
                logging.info('Vehicle not available!!')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def return_rented_vehicle(self, rented_veh):
        try:
            logging.info(f'Returning rented vehicle {rented_veh}')
            self.avl_veh.append(rented_veh)
            logging.info(f'{rented_veh} returned.')
            return f'{rented_veh} returned.'
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

```python
import logging

logging.basicConfig(filename=r'logs\LibraryCatalog.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class LibraryCatalog:

    logging.info('Entering class Library Catalog')

    def __init__(self):
        logging.info('Entering __init__ method. Declaring variable: avl_books')
        self.avl_books = ["To Kill a Mockingbird", "Pride and Prejudice", "The Great Gatsby", 
                          "Harry Potter and the Sorcerer's Stone", "Little Women", "1984", "The Lord of the Rings", 
                          "Alice's Adventures in Wonderland", "The Catcher in the Rye"]

    def add_books(self):
        try:
            opt = str(input('Do you want to add book to the library? Enter "yes" or "no": '))
            logging.info('Taking input from user...')
            if opt.lower() == "yes":
                logging.info('Entering name of the book...')
                new_book = str(input('Enter the name of the book: '))
                self.avl_books.append(new_book)
                print(f'{new_book} added to the library.')
                logging.info(f'{new_book} added to the library.')
            else:
                print('No new book added to the library.')
                logging.info('No new book added to the library.')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def borrow_book(self):
        try:
            opt = str(input('Do you want to borrow book from the library? Enter "yes" or "no": '))
            logging.info('Taking input from user...')
            if opt.lower() == "yes":
                logging.info(f'Showing available books...\n{self.avl_books}')
                print(self.avl_books)
                brw_book = str(input('Enter the name of the book you want to borrow: '))
                if brw_book in self.avl_books:
                    self.avl_books.remove(brw_book)
                    print(f'{brw_book} borrowed from the library.')
                    logging.info(f'{brw_book} borrowed from the library.')
                else:
                    print(f'{brw_book} not available in the library.')
                    logging.info(f'{brw_book} not available in the library.')
            else:
                print('No book borrowed from the library.')
                logging.info('No book borrowed from the library.')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def available_books(self):
        try:
            logging.info(f'Displaying available books...\n{self.avl_books}')
            print(self.avl_books)
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

Problem 5: Product Inventory

```python
import logging

logging.basicConfig(filename=r'logs\ProductInventory.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class ProductInventory:

    logging.info('Entering class Product Inventory.')

    def __init__(self):
        logging.info('Entering __init__ method. Declaring variable: inventory')
        self.inventory = [{"product_id": 101, "name": "Laptop", "category": "Electronics", "price": 75000, "stock": 10},
                          {"product_id": 102, "name": "Smartphone", "category": "Electronics", "price": 40000, "stock": 25},
                          {"product_id": 103, "name": "Headphones", "category": "Accessories", "price": 2000, "stock": 50},
                          {"product_id": 104, "name": "Desk Chair", "category": "Furniture", "price": 5000, "stock": 15},
                          {"product_id": 105, "name": "Coffee Mug", "category": "Kitchen", "price": 500, "stock": 100}]

    def add_product(self):
        try:
            logging.info('Add a new product.\nTaking input from user')
            prod_id, price, stock = map(int, input('Enter the product ID, price and stock of new product seperated by a space:').split())
            name, category = map(str, input('Enter the name and category of new product seperated by a space:').split())
            new_prod = {"product_id": prod_id, "name": name, "category": category, "price": price, "stock": stock}
            logging.info('Adding new product to the inventory.')
            self.inventory.append(new_prod)
            logging.info('New product added to the inventory.')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def update_product(self):
        try:
            logging.info('Update request raised.')
            prod_id = int(input('Enter product_id which needs an update: '))
            for item in self.inventory:
                if item["product_id"] == prod_id:
                    logging.info('Product found. Updating stock quantity.')
                    item["stock"] = int(input('Enter the new quantity: '))
                    logging.info('Stock updated.')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def display_products(self):
        try:
            logging.info('Displaying available products')
            for item in self.inventory:
                print(item["name"], end=' ')
            logging.info(f'{self.inventory}')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

Problem 6: Shape Calculation

```python
import math
import logging

logging.basicConfig(filename=r'logs\ShapeCalculator.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class ShapeCalculator:
    def __init__(self, shape, *args):
        self.shape = shape.lower()
        self.args = args  # Store all input parameters
        logging.info(f'Fetching details from user to calculate area and perimeter of {self.shape}')

        # Calculate area and perimeter
        self.area, self.perimeter = self.calculate_area_perimeter()

        logging.info(f'Area of {self.shape}: {self.area}\nPerimeter of {self.shape}: {self.perimeter}')
        print(f'Area of {self.shape}: {self.area}\nPerimeter of {self.shape}: {self.perimeter}')

    def calculate_area_perimeter(self):
        if self.shape == "circle":
            r = self.args[0]  # Expect radius
            area = math.pi * r ** 2
            perimeter = 2 * math.pi * r
        elif self.shape == "rectangle":
            l, w = self.args  # Expect length and width
            area = l * w
            perimeter = 2 * (l + w)
        elif self.shape == "square":
            s = self.args[0]  # Expect side length
            area = s ** 2
            perimeter = 4 * s
        elif self.shape == "triangle":
            b, h, s1, s2, s3 = self.args  # Expect base, height, and 3 sides
            area = 0.5 * b * h
            perimeter = s1 + s2 + s3
        else:
            raise ValueError("Shape not supported!")

        return area, perimeter
```

Problem 7: Student Management

```python
import logging

logging.basicConfig(filename=r'logs\StudentManagement.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class StudentManagement:

    logging.info('Entering class Student Management.')

    def __init__(self, std_id, name, *grade):
        logging.info('Entering __init__ method. Declaring Variable: std_id, std_name, std_grade')
        self.std_id = std_id
        self.std_name = name
        self.std_grade = grade

    def avg_grade(self):
        try:
            logging.info('Calculating average grade of the student.')
            total_grade = 0
            count = 0
            for i in self.std_grade:
                total_grade += i
                count += 1
            avg_grade = total_grade/count
            logging.info(f'Average Grade: {avg_grade}')
            print(f'Average Grade: {avg_grade}')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
            
    def display_std_details(self):
        try:
            logging.info('Displaying student details.')
            print(f'Student ID: {self.std_id}\nStudent Name: {self.std_name}\nStudent Grades: ')
            logging.info(f'Student ID: {self.std_id}\nStudent Name: {self.std_name}\nStudent Grades: ')
            for i in self.std_grade:
                print(i)
                logging.info(i)
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

Problem 10: ToDo List

```python
import logging

logging.basicConfig(filename=r'logs\ToDoList.log', level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class ToDoList:

    logging.info('Entering class ToDo List.')

    def __init__(self, **tasks):
        logging.info('Entering __init__ method. Declaring variable: tasks')
        self.tasks = tasks

    def add_tasks(self, new_task, due_date, completion_status):
        try:
            logging.info('Adding new task, due date and completion status...')
            self.tasks[new_task] = [due_date, completion_status]
            logging.info('New task added!!')
            print('New task added!!')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def completion_status(self, task):
        try:
            for key, value in self.tasks.items():
                if key == task:
                    self.tasks[key][1] = 'completed'
            logging.info('Task completed!!')
            print('Task completed!!')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')

    def display_pending_tasks(self):
        try:
            logging.info('Pending tasks:')
            for key, value in self.tasks.items():
                if value[1].lower() == 'incomplete':
                    print(f'Task: {key}, Due Date: {self.tasks[key][0]}, Completion Status: {self.tasks[key][1]}')
                    logging.info(f'Task: {key}, Due Date: {self.tasks[key][0]}, Completion Status: {self.tasks[key][1]}')
        except Exception as e:
            logging.error(f'An error has occurred\n{e}')
```

```python
user1 = ToDoList(Print_Documents=['26-03-2025', 'incomplete'], Get_Driving_Liscense=['30-06-2025', 'incomplete'])
```

```python
user1.add_tasks('Get_Passport_Size_Photos', '20-03-2025', 'incomplete')
```

```python
user1.completion_status('Print_Documents')
```

```python
user1.display_pending_tasks()
```
